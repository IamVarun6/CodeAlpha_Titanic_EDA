```python
df['Survived'].value_counts()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.countplot(x='Survived', data=df)

plt.title("Survival Distribution")
plt.show()
```

The survival distribution shows that the majority of passengers did not survive the Titanic disaster. Out of 891 passengers, 549 (61.6%) died and 342 (38.4%) survived.

```python
df['Sex'].value_counts()
```

```python
sns.countplot(x='Sex', data=df)

plt.title("Gender Distribution")
plt.show()
```

```python
pd.crosstab(df['Sex'], df['Survived'])
```

Female passengers had a significantly higher survival rate than male passengers. Approximately 74.2% of females survived, compared to only 18.9% of males. This indicates that gender was a major factor influencing survival.

```python
sns.countplot(
    x='Sex',
    hue='Survived',
    data=df
)

plt.title("Gender vs Survival")
plt.show()
```

```python
df['Pclass'].value_counts()
```

```python
pd.crosstab(df['Pclass'], df['Survived'])
```

Third-class passengers formed the largest group with 491 passengers, followed by first-class (216) and second-class passengers (184).

Passenger class had a significant impact on survival. First-class passengers had the highest survival rate (63.0%), followed by second-class passengers (47.3%). Third-class passengers had the lowest survival rate (24.2%), indicating that passengers in higher classes were more likely to survive.

```python
sns.countplot(
    x='Pclass',
    hue='Survived',
    data=df
)

plt.title("Passenger Class vs Survival")
plt.show()
```

```python
df['Age'].describe()
```

```python
sns.histplot(df['Age'], bins=20, kde=True)
plt.title("Age Distribution")
plt.show()
```

The age distribution peaks between 20 and 30 years, indicating that young adults formed the largest passenger group aboard the Titanic.

The number of passengers gradually decreases beyond the age of 40, suggesting that older individuals represented a smaller portion of the population.

A small number of infants and young children were present in the dataset.

Elderly passengers were relatively uncommon compared to younger adults.

The age distribution is concentrated among young adults and gradually tapers off toward both younger and older age groups.

```python
sns.boxplot(x='Survived', y='Age', data=df)

plt.title("Age vs Survival")
plt.show()
```

The boxplot indicates that survivors tended to be slightly younger than non-survivors. This suggests that age may have influenced survival, although the relationship appears weaker than the effects of gender and passenger class.
